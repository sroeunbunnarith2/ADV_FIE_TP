**Class**: is father of function 

**Variable**: is a named container in programming used to store and manage data in memory. It holds a value—such as a number, text, or boolean—that can be referenced and modified throughout a program's execution

**Loop**: is a control flow structure that repeatedly executes a block of code as long as a specified condition remains true. It is used to automate repetitive tasks without rewriting the same code multiple times (e.g., for loops, while loops)
